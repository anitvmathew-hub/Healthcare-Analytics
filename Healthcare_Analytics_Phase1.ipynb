## Dataset Information

### Dataset Source
The dataset used in this project is the **Diabetes 130-US Hospitals for Years 1999–2008** dataset, obtained from the UCI Machine Learning Repository.

### Dataset Description
The dataset contains historical hospital encounter records for patients with diabetes. It includes information about patient demographics, hospital stays, diagnoses, medical procedures, medications, and readmission outcomes.

Each row represents a hospital encounter rather than a unique patient. A patient may have multiple encounters recorded in the dataset.

### Dataset Details
- **Dataset name:** Diabetes 130-US Hospitals for Years 1999–2008
- **File name:** `diabetic_data.csv`
- **File format:** CSV
- **Number of records:** 101,766
- **Number of columns:** 50
- **Domain:** Healthcare Analytics

### Key Fields
Some important fields for this project include:

- **age** – The patient's age group.
- **gender** – The patient's recorded gender.
- **time_in_hospital** – The number of days spent in the hospital.
- **num_lab_procedures** – The number of laboratory procedures recorded during the encounter.
- **num_procedures** – The number of procedures recorded during the encounter.
- **num_medications** – The number of medications recorded during the encounter.
- **number_inpatient** – The number of inpatient visits recorded in the previous year.
- **diag_1, diag_2, and diag_3** – The primary, secondary, and additional diagnoses recorded for the encounter.
- **readmitted** – Indicates whether the patient was readmitted in less than 30 days, after 30 days, or had no recorded readmission in the dataset's defined categories.

### Project Limitations

There are a few limitations to keep in mind while working on this project.

1. **Historical data:** The dataset covers hospital encounters from 1999 to 2008. Healthcare practices may have changed since then, so the findings may not represent current hospital conditions.

2. **Limited information:** The dataset contains only the information recorded in the available fields. Other factors that may influence readmission, such as a patient's home support or access to follow-up care, may not be fully represented.

3. **Relationships do not prove causes:** If certain characteristics are associated with higher readmission rates, this does not necessarily mean they caused the readmissions.

4. **Readmission is not the whole picture:** Returning to the hospital does not automatically mean that a patient received poor care. Readmission patterns can provide useful information, but they cannot measure overall care quality on their own.

5. **Repeated hospital encounters:** The same patient may appear in multiple records. I will therefore distinguish between hospital encounters and unique patients when interpreting the findings.

These limitations will help me interpret the results carefully and avoid drawing conclusions that the available data cannot support.
