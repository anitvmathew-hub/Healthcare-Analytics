```python
print("=" * 144)
print("                                                       IQR outlier detection")
print("=" * 144)
```

```python
# Select numerical columns, excluding identifiers
numeric_columns = df_cleaned.select_dtypes(include=["int64", "float64"]).columns.drop(["encounter_id", "patient_nbr"])

# Identify potential outliers using IQR
for column in numeric_columns:
    Q1 = df_cleaned[column].quantile(0.25)
    Q3 = df_cleaned[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df_cleaned[(df_cleaned[column] < lower_limit) |(df_cleaned[column] > upper_limit)]

    print(f"{column}: {len(outliers)} potential outliers")
```

Outlier Detection Using IQR

* The Interquartile Range (IQR) method was applied to identify potential outliers in numerical variables.
* Identifier columns such as encounter_id and patient_nbr were excluded because their values do not represent measurable patient characteristics.
* Potential outliers were identified in several healthcare utilization variables.
* These observations were retained because they may represent genuine patient experiences. 
* No outlier removal or capping was performed.

### *Mapping, Verifying and Correcting Categorical ID Columns*

* The healthcare dataset contains numerical identifiers for admission type, discharge disposition, and admission source. 
* These IDs were difficult to interpret directly, so the hospital ID mapping file was used to convert them into meaningful descriptions.

*Mapping Process:*
- Loaded the hospital ID mapping file.
- Identified and separated the three mapping sections: Admission Type, Discharge Disposition, and Admission Source.
- Created separate mapping DataFrames for each category.
- Used Pandas `merge()` with a left join to add the corresponding descriptions to the analysis dataset.
- Preserved the original numerical ID columns for reference and further analysis.

*Verification and Correction:*
- Verified that the mapped descriptions matched their corresponding ID values.
- Checked the data types of the ID columns to ensure compatibility during merging.
- Identified missing descriptions in the admission type mapping.
- Replaced missing admission type descriptions with `"Unknown"` to improve readability and consistency.
- Verified the mapping results using sample records and frequency counts.

*Outcome:*

* The three categorical ID columns were successfully enriched with meaningful descriptions.
* This makes the dataset easier to interpret and supports more meaningful analysis of hospital admission patterns, discharge outcomes, and admission sources.
* The original `df_cleaned` dataset was preserved, while the additional descriptive columns were added to the separate `df_analysis` dataset.

```python
# Admission Type Mapping
admission_mapping = mapping_df.iloc[0:8].copy()
admission_mapping.columns = ["admission_type_id", "admission_type"]

# Discharge Disposition Mapping
discharge_mapping = mapping_df.iloc[10:40].copy()
discharge_mapping.columns = ["discharge_disposition_id", "discharge_disposition"]

# Admission Source Mapping
source_mapping = mapping_df.iloc[42:67].copy()
source_mapping.columns = ["admission_source_id", "admission_source"]

# Convert ID columns to numeric
admission_mapping["admission_type_id"] = pd.to_numeric(admission_mapping["admission_type_id"])
discharge_mapping["discharge_disposition_id"] = pd.to_numeric(discharge_mapping["discharge_disposition_id"])
source_mapping["admission_source_id"] = pd.to_numeric(source_mapping["admission_source_id"])

# Remove rows with missing IDs
admission_mapping = admission_mapping.dropna(subset=["admission_type_id"])
discharge_mapping = discharge_mapping.dropna(subset=["discharge_disposition_id"])
source_mapping = source_mapping.dropna(subset=["admission_source_id"])

# Merging all three mapping DataFrames
df_analysis = df_analysis.drop(columns=["admission_type", "discharge_disposition", "admission_source"],errors="ignore")
df_analysis = df_analysis.merge(admission_mapping,on="admission_type_id",how="left")

df_analysis = df_analysis.merge(discharge_mapping,on="discharge_disposition_id",how="left")

df_analysis = df_analysis.merge(source_mapping,on="admission_source_id",how="left")
```
